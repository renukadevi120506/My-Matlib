```python
import matplotlib.pyplot as plt
x=(1,2,3,4,5,6)
y=(2,4,6,8,10,12)


plt.plot(x, y)
plt.title("Simple plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

```

```python
import matplotlib.pyplot as plt
a=[2,3,4,5,6,7]
b=[4,9,16,25,36,49]

plt.scatter(a,b ,color='Green' , marker='o')
plt.title("Scatter Plot Example")
plt.xlabel("A-Axis")
plt.ylabel("B-Axis")
plt.show()
```

```python
a =['apple','orange','grapes','banana','watermelon']
b =[2,4,6,7,9]

plt.bar(a,b ,color ='red')
plt.title("Mat Bar Example")
plt.xlabel("Fruits Name")
plt.ylabel("No of fruits")
plt.show()
```

```python
a=['Python','Java','JavaScript','CSS','PHP']
b=[89,98,70,99,80]

plt.barh(a,b , color='orange')
plt.title("My Programming Score")
plt.xlabel("programming language")
plt.ylabel("Score")
plt.show()
```

```python
import numpy as np

nd=np.random.randn(2000)

plt.hist(nd, bins=120,color='pink',edgecolor='black')
plt.title("My Numpy")
plt.xlabel("Value")
plt.ylabel("frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
sizes =[10,15,25,50]
labels=['Spirit','Coke','Monster','Redbull']

plt.pie(sizes,labels=labels,autopct='%1.1f%%', startangle=140)
plt.title("Soft Drinking")
plt.show()
```

```python
data=[24,56,78,35,78,90]

plt.boxplot(data)
plt.title("Box value")
plt.ylabel("Value")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y1=[2,3,4,6,8,10,12]
y2=[1,3,5,7,9,11,13]

plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("plot 1")

plt.subplot(1,2,2)
plt.bar(x,y2)
plt.title("plot 2")

plt.show()
```

```python
x=[1,2,3,4]

plt.plot(x,[3,5,7,9],label='square')
plt.plot(x,[2,4,5,6],label='cube')

plt.legend()
plt.title("Legend Example")
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]
error = [0.2, 0.4, 0.6, 0.5]

plt.errorbar(x, y, yerr=error, fmt='o', color='black', ecolor='red', capsize=5)
plt.title("Error Bar Example")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y,color='red')
plt.fill_between(x, y, color='black', alpha=0.4)
plt.title("Fill Between Example")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y,color='green')
plt.annotate("Peak Point", xy=(5, 10), xytext=(4, 9),
             arrowprops=dict(facecolor='blue', shrink=0.09))
plt.show()
```
